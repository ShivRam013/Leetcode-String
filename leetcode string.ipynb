```python
#Reverse String
```

```python
s = ["h","e","l","l","o"]
#Output: ["o","l","l","e","h"]
```

```python
def rev(s):
    left = 0
    right = len(s)-1
    while left < right:
        s[left],s[right] = s[right],s[left]
        left+=1
        right-=1
    return s


rev(s)
```

```python
s = "A man, a plan, a canal: Panama"
#Output: true
```

```python
def rev(s):
    new = []
    for i in s:
        if i.isalnum():
            new.append(i.lower())
    return new == new[::-1]
        
        
rev(s)
```

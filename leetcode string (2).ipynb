```python
#Reverse String
```

```python
s = ["h","e","l","l","o"]
#Output: ["o","l","l","e","h"]
```

```python
def rev(s):
    left = 0
    right = len(s)-1
    while left < right:
        s[left],s[right] = s[right],s[left]
        left+=1
        right-=1
    return s


rev(s)
```

```python
#Valid Palindrome
```

```python
s = "A man, a plan, a canal: Panama"
#Output: true
```

```python
def rev(s):
    new = []
    for i in s:
        if i.isalnum():
            new.append(i.lower())
    return new == new[::-1]
        
        
rev(s)
```

```python
#Valid Anagram
```

```python
s = "anagram"
t = "nagaram"
#Output: true
```

```python
def isAnagram(s,t):
    if len(s) != len(t):
        return False

    counter = {}

    for char in s:
        counter[char] = counter.get(char, 0) + 1

    for char in t:
        if char not in counter or counter[char] == 0:
            return False
        counter[char] -= 1

    return True


isAnagram(s,t)
```

```python
#First Unique Character in a String
```

```python
s = "loveleetcode"
#Output: 2
```

```python
def uni(s):
    counter = {}
    for x in s:
        counter[x] = counter.get(x,0)+1
    for x in s:
        if counter[x] == 1:
            return s.index(x)
    return -1

uni(s)
```

```python
#Longest Common Prefix
```

```python
strs = ["flower","flow","flight"]
#Output: "fl"
```

```python
def longestCommonPrefix(v):
    ans=""
    v=sorted(v)
    first=v[0]
    last=v[-1]
    for i in range(min(len(first),len(last))):
        if first[i]!=last[i]:
            return ans
        ans+=first[i]
    return ans


longestCommonPrefix(strs)
```

```python
#Find the Index of the First Occurrence in a String
```

```python
haystack = "shadbutsad"
needle = "sad"
#Output: 7
#Explanation: "sad" occurs at index 0 and 6.
```

```python
def strStr(haystack, needle):
    for i in range(len(haystack) - len(needle) + 1):
        if haystack[i:i+len(needle)] == needle:
            return i
    return -1

strStr(haystack,needle)
#haystack[7:10]
```

```python
s = "([])"

#Output: true
```

```python
def isValid(s):
    
    while '()' in s or '[]'in s or '{}' in s:
        s = s.replace('()','').replace('[]','').replace('{}','')
    return False if len(s) !=0 else True
    
    
    
isValid(s)
```
